```python
import pandas as pd, numpy as np
```

```python
df = pd.read_csv("../data/raw/user.csv")
df.head()
```

```python
df.info()
```

```python
# number of null value
print(f"Number of Null Value: \n{df.isna().sum()}")

# number of duplicate value
print(f"\nNumber of duplicate value: {df.duplicated().sum()}")
```

```python
users = df.copy()

users["signup_date"] = pd.to_datetime(users["signup_date"])
users["acquisition_channel"] = users["acquisition_channel"].str.strip().str.title()
users["city"] = users["city"].str.strip().str.title()
users["age_missing"] = users["age"].isna().astype(int)  # flag age column for marking missing value
```

```python
users.info()
```

```python
users.to_csv("../data/clean/user_clean.csv", index=False)
```
